```python
# Electricity Bill Calculator

name = input("Enter customer name: ")
units = float(input("Enter electricity units consumed: "))

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = units * 3
elif units <= 300:
    bill = units * 5
else:
    bill = units * 7

print("\nCustomer Name:", name)
print("Units Consumed:", units)
print("Electricity Bill: ₹", bill)
```

```python
# Electricity Bill Calculator using User-Defined Function

def calculate_bill(units):
    if units <= 100:
        bill = units * 2
    elif units <= 200:
        bill = units * 3
    elif units <= 300:
        bill = units * 5
    else:
        bill = units * 7

    return bill


name = input("Enter customer name: ")
units = float(input("Enter electricity units consumed: "))

bill = calculate_bill(units)

print("\nCustomer Name:", name)
print("Units Consumed:", units)
print("Electricity Bill: ₹", bill)
```

### Scenario: Basic Math Operations Module

Let's create a module named `math_operations.py` that will contain functions for addition, subtraction, multiplication, and division. Then, we will import this module and use its functions.

```python
%%writefile math_operations.py

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Cannot divide by zero!"
    return a / b
```

Now that the `math_operations.py` module is created, let's import it and use its functions.

```python
import math_operations

num1 = 10
num2 = 5

print(f"Addition: {num1} + {num2} = {math_operations.add(num1, num2)}")
print(f"Subtraction: {num1} - {num2} = {math_operations.subtract(num1, num2)}")
print(f"Multiplication: {num1} * {num2} = {math_operations.multiply(num1, num2)}")
print(f"Division: {num1} / {num2} = {math_operations.divide(num1, num2)}")

num3 = 10
num4 = 0
print(f"Division: {num3} / {num4} = {math_operations.divide(num3, num4)}")
```
