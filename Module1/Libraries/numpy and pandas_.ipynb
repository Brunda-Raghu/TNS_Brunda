```python
# Iterating through Rows
arr1 = np.arange(30).reshape(5,6)  
print("Printing Through Rows:")
for row in arr1:      # Returns one row at a time - default
    print(row)
# Iterating through Rows
print("Printing Through Columns:")
cols = arr1.shape[1]   # Returns number of columns
for n in range(cols):
    print(arr1[:,n])  # Prints all elements on n-th column
```

```python
# Horizontal Stacking of Two Arrays
arr1 = np.arange(30).reshape(5,6)
print("Array 1 :\n", arr1)
arr2 = np.arange(100, 130).reshape(5,6)
print("Array 2 :\n", arr2)
print("Stacked Array :\n", np.hstack((arr1, arr2)))
```

```python
# Vertical Stacking of Two Arrays
arr1 = np.arange(30).reshape(5,6)
print("Array 1 :\n", arr1)
arr2 = np.arange(100, 130).reshape(5,6)
print("Array 2 :\n", arr2)
print("Stacked Array :\n", np.vstack((arr1, arr2)))
```

```python
# Vertical Splitting of Two Arrays
arr1 = np.arange(36).reshape(6,6)
print(arr1, '\n')
splt = np.vsplit(arr1, 2)
print(splt[0], '\n')
print(splt[1])
```

```python
# Horizontal Splitting of Two Arrays
arr1 = np.arange(36).reshape(6,6)
print(arr1, '\n')
splt = np.hsplit(arr1, 2)
print(splt[0], '\n')
print(splt[1])
```

```python
# Demonstrate Array assignment
arr1 = np.arange(36).reshape(6,6)
arr2 = arr1       # arr2 is a reference to arr1, and not a physically different copy
arr1[1,1] = 99    # Change reflected in both arr1 and arr2
print("Array 1: \n", arr1)
print("Array 2: \n", arr2)
```

```python
# Demostrate Array Copy
arr1 = np.arange(36).reshape(6,6)
arr2 = arr1.copy()  # A physically different copy of the input array is created 
arr1[1,1] = 99      # Change reflected in both arr1 and arr2
print("Array 1: \n", arr1)
print("Array 2: \n", arr2)
```

```python
# Copies one Series to a new one.
# Deep copy: Creates a new object with values and indexes of the original Series. Change in any element in the 
# original Series will not reflect in the Deep copy.
# Shallow Copy: Creates a new object with references of values and indexes of the original Series. 
# Change in any element in the original Series will reflect in the Shallow copy.

# Series Deep Copy & Shallow Copy
ser1 = pd.Series([3, 4, 5])
ser2 = ser1.copy(deep=True)
ser1[2] = 8
print("Ser1: \n", ser1)
print("Ser2 - Deep Copy: \n", ser2)
ser3 = ser1.copy(deep=False)
ser1[2] = 8
print("Ser3 - Shallow Copy: \n", ser3)
```

```python
# In the folowing code, we will perform an operation to find the cube of a given number.
# To do that, we will use three different methods:
# - Use a traditional Python function created using 'def'
# - Create a named Lambda function
# - Use an Inline Lambda function

from time import time_ns

def cubed(x):
        return x**3

lambda_cubed = lambda x:x**3

start = time_ns()   # Returns time in nanoseconds since epoch (Jan 1, 1970) 

print(f"Cube of 9 is {cubed(9)}")
print(f"Time taken by traditional method : {(time_ns()-start)//1000} ms\n")

start = time_ns()

print(f"Cube of 9 is {lambda_cubed(9)}")
print(f"Time taken by named lambda function: {(time_ns()-start)//1000} ms\n")

start = time_ns()
print(f"Cube of 9 is {(lambda x:x**3)(9)}")
print(f"Time taken by inline lambda function: {(time_ns()-start)//1000} ms")
```

```python
# E.g. We are to return all number greater than 5 from a list of numbers

numbers = [4, 3, 6, 7, 3, 9, 11]

def filt_gt5(num):
    return num < 5
```

```python
filt = filter(filt_gt5, numbers)
list(filt)
```

```python
# Can also be inline filter using a Lambda function

filt = filter(lambda x: x > 5, numbers)
list(filt)
```
