```python
#LIST
#Problem Statement
#An online shopping application needs to store the products selected by a
#customer. The customer can add new products, remove products, and modify the
#cart.

#Why choose List?

#A list is suitable because:

#It is ordered.
#It is mutable, so items can be added or removed.
#Duplicate products are allowed.
#We can access products using their index

# Shopping Cart using List

cart = ["Laptop", "Mouse", "Keyboard"]

print("Initial Cart:")
print(cart)

# Add products
cart.append("Headphones")
cart.append("USB Cable")

print("\nAfter adding products:")
print(cart)

# Remove a product
cart.remove("Mouse")

print("\nAfter removing Mouse:")
print(cart)

# Change a product
cart[1] = "Wireless Keyboard"

print("\nAfter modifying the cart:")
print(cart)

print("\nTotal products:", len(cart))
```

```python
#TUPLE
#Problem Statement

#A navigation application needs to store the latitude and longitude of a
#particular location. These coordinates should not accidentally be changed.

#Why choose Tuple?

#A tuple is suitable because:

#It is ordered.
#It is immutable.
#The latitude and longitude should remain fixed.
#It can store multiple related values together

# GPS Location using Tuple

location = (12.9716, 77.5946)

print("GPS Coordinates:")
print("Latitude:", location[0])
print("Longitude:", location[1])

print("\nComplete Location:", location)
print("Number of coordinates:", len(location))
```

```python
#SET
#Problem Statement

#A newsletter system collects email addresses from users. The same person may
#subscribe more than once, but the system should store each email address only
#once.

#Why choose Set?

#A set is suitable because:

#It stores unique values.
#Duplicate values are automatically removed.
#It supports operations such as union, intersection, and difference.
#Order is not important for subscribers.


# Newsletter Subscribers using Set

subscribers = {
    "rahul@gmail.com",
    "priya@gmail.com",
    "arun@gmail.com",
    "rahul@gmail.com"
}

print("Unique Subscribers:")
print(subscribers)

# Add a new subscriber
subscribers.add("sneha@gmail.com")

print("\nAfter adding a subscriber:")
print(subscribers)

# Try adding duplicate email
subscribers.add("priya@gmail.com")

print("\nAfter adding duplicate email:")
print(subscribers)

# Remove a subscriber
subscribers.remove("arun@gmail.com")

print("\nAfter removing a subscriber:")
print(subscribers)

print("\nTotal unique subscribers:", len(subscribers))
```

```python
#DICTIONARY
#Problem Statement

#A company needs to store employee information where each employee has an
#employee ID, name, department, and salary. The information should be
#accessed using the employee ID.

#Why choose Dictionary?

#A dictionary is suitable because:

#Data is stored as key-value pairs.
#Employee ID can be used as a unique key.
#Employee details can be accessed quickly using the key.
#Multiple details can be stored for each employee.


# Employee Details using Dictionary

employees = {
    101: {
        "name": "Rahul",
        "department": "IT",
        "salary": 45000
    },
    102: {
        "name": "Priya",
        "department": "HR",
        "salary": 40000
    },
    103: {
        "name": "Arun",
        "department": "Finance",
        "salary": 50000
    }
}

# Display employee details
print("Employee 101 Details:")
print("Name:", employees[101]["name"])
print("Department:", employees[101]["department"])
print("Salary:", employees[101]["salary"])

# Add a new employee
employees[104] = {
    "name": "Sneha",
    "department": "Marketing",
    "salary": 42000
}

print("\nAfter adding new employee:")
print(employees)

# Update salary
employees[101]["salary"] = 48000

print("\nUpdated salary of Employee 101:")
print(employees[101]["salary"])
```
