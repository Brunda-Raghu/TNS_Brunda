```python
# Electricity Bill Calculator

name = input("Enter customer name: ")
units = float(input("Enter electricity units consumed: "))

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = units * 3
elif units <= 300:
    bill = units * 5
else:
    bill = units * 7

print("\nCustomer Name:", name)
print("Units Consumed:", units)
print("Electricity Bill: ₹", bill)
```

```python
# Electricity Bill Calculator using User-Defined Function

def calculate_bill(units):
    if units <= 100:
        bill = units * 2
    elif units <= 200:
        bill = units * 3
    elif units <= 300:
        bill = units * 5
    else:
        bill = units * 7

    return bill


name = input("Enter customer name: ")
units = float(input("Enter electricity units consumed: "))

bill = calculate_bill(units)

print("\nCustomer Name:", name)
print("Units Consumed:", units)
print("Electricity Bill: ₹", bill)
```
